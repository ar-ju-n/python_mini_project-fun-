```python
import time
import cv2
import numpy as np
from mss import mss


with mss() as sct:

    monitor = sct.monitors[1]

    width = monitor["width"]
    height = monitor["height"]
    target_resolution = (width, height)

    codec = cv2.VideoWriter_fourcc(*"XVID")  # type: ignore
    filename = "Recording.avi"
    fps = 30.0
    frame_delay = 1.0 / fps

    out = cv2.VideoWriter(filename, codec, fps, target_resolution)

    cv2.namedWindow("Live", cv2.WINDOW_NORMAL)
    cv2.resizeWindow("Live", 480, 270)

    print(f"Recording screen at actual size: {width}x{height}")
    print(f"Press 'q' in the prview window to stop.")

    while True:
        start_time = time.time()

        scr = sct.grab(monitor)

        frame = np.array(scr)

        frame = cv2.cvtColor(frame, cv2.COLOR_BGRA2BGR)

        out.write(frame)

        cv2.imshow("Live", frame)

        if cv2.waitKey(1) == ord('q'):
            break

        elapsed_time = time.time() - start_time
        if elapsed_time < frame_delay:
            time.sleep(frame_delay - elapsed_time)

out.release()
cv2.destroyAllWindows()
print("Recording saved successfully")
```
