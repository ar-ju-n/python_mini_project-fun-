```python
# importing required libraries
import sounddevice as sd
import wavio as wv
from scipy.io.wavfile import write
```

```python

# Sampling frequecies and duration of recording
freq = 44100

duration = 5

# Start recorder with the given values fo duration and sample frequency
recording = sd.rec(int(duration * freq), samplerate=freq, channels=2)

# record audio for the given number of seconds
sd.wait()

# Coonverting the Numpy array to an audio files with the given sampling frequency
write("recording0.wav", freq, recording)


# convert the numpy array to audio file
wv.write("recording2.wav", recording, freq, sampwidth=2)
```
